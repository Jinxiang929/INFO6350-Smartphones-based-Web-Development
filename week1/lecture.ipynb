```python
import tkinter as tk

# Create the main window
root = tk.Tk()
root.title("Sum Calculator")
root.geometry("300x150")

# Original calculation
num1 = 10
num2 = 20
result = num1 + num2

# Display the result
label = tk.Label(
    root,
    text=f"The sum is: {result}",
    font=("Arial", 16)
)
label.pack(pady=40)

# Start the application
root.mainloop()
```

```python
import tkinter as tk

def add_numbers():
    num1 = float(entry1.get())
    num2 = float(entry2.get())
    result = num1 + num2
    result_label.config(text=f"The sum is: {result}")

# Create window
root = tk.Tk()
root.title("Sum Calculator")
root.geometry("300x250")

# First number
label1 = tk.Label(root, text="Enter first number:")
label1.pack()

entry1 = tk.Entry(root)
entry1.pack()

# Second number
label2 = tk.Label(root, text="Enter second number:")
label2.pack()

entry2 = tk.Entry(root)
entry2.pack()

# Add button
button = tk.Button(root, text="Add", command=add_numbers)
button.pack(pady=20)

# Result
result_label = tk.Label(root, text="The sum is:")
result_label.pack()

# Run application
root.mainloop()
```
