```python

import tkinter as tk
from tkinter import messagebox, filedialog
import qrcode
from PIL import Image, ImageTk


# -----------------------------------------
# QR CODE GENERATOR APPLICATION
# -----------------------------------------

class QRCodeApp:

    def __init__(self, root):

        self.root = root
        self.root.title("QR Code Generator")
        self.root.geometry("500x600")
        self.root.resizable(False, False)

        self.qr_image = None
        self.preview_image = None

        # ---------------------------------
        # Title
        # ---------------------------------
        title = tk.Label(
            root,
            text="QR Code Generator",
            font=("Arial", 20, "bold")
        )
        title.pack(pady=20)

        # ---------------------------------
        # Input Label
        # ---------------------------------
        label = tk.Label(
            root,
            text="Enter Text or URL:",
            font=("Arial", 12)
        )
        label.pack()

        # ---------------------------------
        # Text Entry
        # ---------------------------------
        self.text_entry = tk.Entry(
            root,
            width=45,
            font=("Arial", 12)
        )
        self.text_entry.pack(pady=10)

        # ---------------------------------
        # Generate Button
        # ---------------------------------
        generate_button = tk.Button(
            root,
            text="Generate QR Code",
            command=self.generate_qr,
            font=("Arial", 12, "bold"),
            width=20
        )
        generate_button.pack(pady=10)

        # ---------------------------------
        # QR Code Display Area
        # ---------------------------------
        self.qr_label = tk.Label(root)
        self.qr_label.pack(pady=20)

        # ---------------------------------
        # Save Button
        # ---------------------------------
        save_button = tk.Button(
            root,
            text="Save QR Code",
            command=self.save_qr,
            font=("Arial", 12),
            width=20
        )
        save_button.pack(pady=10)

        # ---------------------------------
        # Status Label
        # ---------------------------------
        self.status_label = tk.Label(
            root,
            text="Enter text or a URL to generate a QR code.",
            font=("Arial", 10)
        )
        self.status_label.pack(pady=10)

    # -----------------------------------------
    # Generate QR Code
    # -----------------------------------------
    def generate_qr(self):

        text = self.text_entry.get().strip()

        # Check if the input is empty
        if text == "":
            messagebox.showwarning(
                "Input Required",
                "Please enter text or a URL."
            )
            return

        try:

            # Create QR Code
            qr = qrcode.QRCode(
                version=None,
                error_correction=qrcode.constants.ERROR_CORRECT_H,
                box_size=10,
                border=4
            )

            qr.add_data(text)
            qr.make(fit=True)

            # Create QR image
            self.qr_image = qr.make_image(
                fill_color="black",
                back_color="white"
            ).convert("RGB")

            # Resize image for GUI display
            preview = self.qr_image.resize(
                (250, 250)
            )

            self.preview_image = ImageTk.PhotoImage(preview)

            # Display QR Code
            self.qr_label.config(
                image=self.preview_image
            )

            self.status_label.config(
                text="QR Code generated successfully!"
            )

        except Exception as e:

            messagebox.showerror(
                "Error",
                str(e)
            )

    # -----------------------------------------
    # Save QR Code
    # -----------------------------------------
    def save_qr(self):

        if self.qr_image is None:

            messagebox.showwarning(
                "QR Code Missing",
                "Please generate a QR Code first."
            )
            return

        # Ask user where to save the image
        file_path = filedialog.asksaveasfilename(
            defaultextension=".png",
            filetypes=[
                ("PNG Image", "*.png"),
                ("JPEG Image", "*.jpg"),
                ("All Files", "*.*")
            ],
            title="Save QR Code"
        )

        if file_path:

            self.qr_image.save(file_path)

            messagebox.showinfo(
                "Saved",
                "QR Code saved successfully!"
            )


# -----------------------------------------
# MAIN PROGRAM
# -----------------------------------------

if __name__ == "__main__":

    root = tk.Tk()

    app = QRCodeApp(root)

    root.mainloop()
```
